```python
# --- 1. Subir archivo ---
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

uploaded = files.upload()

# --- 2. Leer archivo con el nombre correcto ---
filename = list(uploaded.keys())[0]   # Toma el nombre exacto del archivo subido
df_games = pd.read_csv(filename)

# --- 5. Gráfico 3: Cantidad de juegos por año de lanzamiento ---
df_games["Year"] = pd.to_datetime(df_games["Launch_date"], errors="coerce").dt.year
df_games["Year"].value_counts().sort_index().plot(kind="bar", figsize=(15,6), color="skyblue")
plt.title("Cantidad de juegos por año")
plt.xlabel("Año")
plt.ylabel("Cantidad")
plt.show()
```

```python
# --- 1. Subir archivo ---
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt

uploaded = files.upload()

# --- 2. Leer archivo con el nombre correcto ---
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)


# --- Tabla cruzada: violencia física vs género ---
tabla = pd.crosstab(df["genero_m_f_victima"], df["fisica"])

# --- Gráfico ---
tabla.plot(kind="bar", figsize=(7,5), color=["skyblue","salmon"])
plt.title("Casos de violencia física por género")
plt.xlabel("Género de la víctima")
plt.ylabel("Cantidad de casos")
plt.legend(title="¿Violencia física?")
plt.xticks(rotation=0)
plt.show()


```
